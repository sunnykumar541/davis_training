```python
import numpy as np

# Create a NumPy array (e.g., numbers from 0 to 9)
numbers = np.arange(10)

# Filter for even numbers
even_numbers = numbers[numbers % 2 == 0]

print("Original array:", numbers)
print("Even numbers:", even_numbers)
```
